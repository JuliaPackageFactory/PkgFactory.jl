# Create a Julia package with PkgFactory

Preview the package files, then create its GitHub repository. Requires Julia 1.12 or later and a GitHub token. Documentation deploy keys use the automatically installed OpenSSH_jll dependency; no separate OpenSSH installation or PATH configuration is needed.

```julia
import Pkg
Pkg.activate(".")
Pkg.add(url="https://github.com/JuliaPackageFactory/PkgFactory.jl.git")
```

```julia
import PkgFactory
```

```julia
spec = PkgFactory.PackageSpec(
    owner = "octocat",
    name = "MyPkg",
    authors = ["The Octocat"],
    description = "A package created from Jupyter",
)
```

```julia
plan = PkgFactory.plan_package(spec)
display(plan)
files = Dict(plan.contents)
println(files["Project.toml"])
```

Review the preview before continuing; the default visibility is public. Set GITHUB_TOKEN in the notebook process's environment. The token needs access to create repositories, write files and workflows, and configure deploy keys and repository secrets. The final cell creates the repository.

```julia
credential = PkgFactory.Credential(ENV["GITHUB_TOKEN"])
```

```julia
result = PkgFactory.create_package(credential, plan)
println(result["url"])
```
